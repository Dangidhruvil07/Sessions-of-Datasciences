```python

import re
import pandas as pd
import numpy as np

def find_contact_details(text: str) -> tuple:

    # Check if the value is missing (NaN)
    if pd.isna(text):
        return np.nan, np.nan

    # Create a regex pattern to identify phone numbers
    # Supports numbers like 951-719-9170 and +1 (217) 569-3204
    phone_number_pattern = re.compile(
        r'(\+?1[\s-]*)?(\(?\d{3}\)?[\s-]*\d{3}[\s-]*\d{4})'
    )

    # Search for the first phone number in the text
    phone_number = re.search(phone_number_pattern, text)

    # Check whether a phone number was found
    if phone_number:
        # Extract the complete matched phone number
        phone_number = phone_number.group().strip()
    else:
        # If no phone number is found, assign NaN
        phone_number = np.nan

    # Remove the phone number from the original text
    # The remaining text should contain the email address
    possible_email_add = re.sub(
        phone_number_pattern, "", text
    ).strip()

    # Return phone number and remaining email text
    return phone_number, possible_email_add
```

```python
# Apply the function to every row of the contact column
# pd.Series converts the returned tuple into two separate columns
patients_df[['phone', 'email']] = patients_df['contact'].apply(
    lambda x: pd.Series(find_contact_details(x))
)
```

```python
treatments_df=treatments_df[treatments_df['dosage_range'] != '-']
```

```python
treatments_df['dosage_start'] = treatments_df['dosage_range'].str.split('-').str.get(0)
treatments_df['dosage_end'] = treatments_df['dosage_range'].str.split('-').str.get(1)
```

```python
treatments_df.drop(columns='dosage_range',inplace=True)
```

```python
treatments_df['dosage_start']=treatments_df['dosage_start'].str.replace('u','')
treatments_df['dosage_end']=treatments_df['dosage_end'].str.replace('u','')
```
