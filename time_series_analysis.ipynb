# Resampling
Resampling involves changing the frequancy of your time series observations.

Two types of resampling

Upresampling :- Where you increase the frequancy of the samples,such as from minitues to second

Downsampling :- Where you decrease the frequancy of the samples as from days to months

```python
# Downsampling
google['Close'].plot()
google['Close'].resample('ME').max().plot()
```

```python
# Upsampling
google['Close'].plot()
google['Close'].resample('H').interpolate(method='spline',order=2).plot()
```

# Rolling window(Smoothing)

Time series data in orignal can be quite and volitile ,espacially on smaller aggregation levels. the concept of rolling or moving avarage is a useful technique for smoothing time series data.

```python
# Rolling Window
# google['Close'].plot()
# google['Close'].rolling(60).mean().plot()
google['Close'].ewm(60).mean().plot()
```
