```python
import seaborn as sns
import numpy as np
```

```python
# import datasets
tips = sns.load_dataset("tips")
iris = sns.load_dataset("iris")
```

# Box plot
a boxplot is a stadardized way of displaying the distribution of data based on five number summary("minimum",first quertile[Q1],meadian,third quartile[Q3] and "maximum").It can tell you about your outliers and what thier values are.boxplots can also tell you if your data is symmetrical, how tightly your data is grouped and if and how your data is skewed



```python
#Box plot
sns.boxplot(data=tips,x='day',y='total_bill')
```

```python
#using catplot figure level
sns.catplot(data=tips,x='day',y='total_bill',kind='box')
```

```python
#using catplot figure level, ,hue
sns.catplot(data=tips,x='day',y='total_bill',kind='box',hue='sex')
```

```python
#sigle boxplot -> numarical col
sns.boxplot(data=tips,y='total_bill')
```

#Categorical Estimate Plot-> central tendency

```python
#barplot
sns.barplot(data=tips,x='sex',y='total_bill',errorbar=None)
```

```python
#catplot
sns.catplot(data=tips,x='sex',y='total_bill',errorbar=None,kind='bar',hue='smoker',estimator=np.std)
```

```python
#catplot
sns.catplot(data=tips,x='sex',y='total_bill',errorbar=None,kind='bar',hue='smoker',estimator='mean')
```

```python
#pointplot
sns.pointplot(data=tips,x='sex',y='total_bill',estimator='mean',hue='smoker',errorbar=None)
```

A special case for the bar plot is when you want to show the number of observations in each category rather than computing a statistic for a
second variable. This is similar to a histogram over a categorical, rather than quantitative, variable

```python
#count plot
sns.countplot(data=tips,x='sex',hue='day')
```

```python
#faceting using catplot
sns.catplot(data=tips,x='sex',y='total_bill',col='smoker',row='time',kind='box')
```

# pairwise Relationship(pairGrid vs pairplot)

```python
sns.pairplot(iris,hue='species')
```

```python
g = sns.PairGrid(data=iris,hue='species')
g.map(sns.scatterplot)
```

```python
# map_diag-> map_offciag
g = sns.PairGrid(data=iris)
g.map_diag(sns.kdeplot)
g.map_offdiag(sns.histplot)
```

```python
 #map_dig -> map_upper - > map.lower
g = sns.PairGrid(data=iris)
g.map_diag(sns.kdeplot)
g.map_upper(sns.histplot)
g.map_lower(sns.scatterplot)
```

```python
# vars
g = sns.PairGrid(data=iris,vars=['sepal_width','petal_width'],hue='species')
g.map_diag(sns.kdeplot)
g.map_upper(sns.histplot)
g.map_lower(sns.scatterplot)
```

```python
# joinplot vs joint grid
```

```python
sns.jointplot(data=tips,x='total_bill',y='tip',kind='hist',hue='sex')
```

```python
# Jointgrid
g = sns.JointGrid(data=tips)
g.plot(sns.scatterplot,sns.histplot)
```

```python
g = sns.JointGrid(data=tips,x='total_bill',y='tip')
g.plot(sns.kdeplot,sns.histplot)
```
