# Annotations

```python
batters = pd.read_csv('https://drive.google.com/uc?export=download&id=1qX0lllLhkXtcv_NeXtDxcSCuy3E_GTfZ')
batters.head()
```

```python
sample_df = batters.head(100).sample(25,random_state=5)
sample_df.head()
```

```python
sample_df['batter'].values[0]
```

```python
plt.figure(figsize=(15,7.5))
plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])

for i in range(sample_df.shape[0]):
  plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

```python
x = [1,2,3,4]
y = [5,6,7,8]
plt.scatter(x,y)
plt.text(1,5,'point 1')
plt.text(2,6,'point 2')
plt.text(3,7,'point 3')
plt.text(4,8,'point 4')
```

# Horizonatal and Vertical

```python
plt.figure(figsize=(15,7.5))
plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])

plt.axhline(130,color='red')
plt.axhline(140,color='green')
plt.axvline(25,color='red')
plt.axvline(35,color='green')

for i in range(sample_df.shape[0]):
  plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
plt.xlabel('Average')
plt.ylabel('Strike rate')
```

# Subplot

```python
# A diff way to plot graphs
plt.scatter(batters['avg'],batters['strike_rate'])
```

```python
fig, ax = plt.subplots(figsize=(15,7.5))
ax.scatter(batters['avg'],batters['strike_rate'],color='red')
ax.set_title('Batsman')
ax.set_xlabel('Average')
ax.set_ylabel('Strike rate')
```

```python
# better dataset
fig, ax = plt.subplots(nrows=2,ncols=1,figsize=(10,5),sharex=True)

ax[0].scatter(batters['avg'],batters['strike_rate'],color='red')
ax[1].scatter(batters['avg'],batters['runs'])

ax[0].set_title('Avarage Vs Strike rate')
ax[0].set_ylabel('Strike rate')

ax[1].set_title('Avarage Vs Runs')
ax[1].set_ylabel('Runs')
ax[1].set_xlabel('Average')
```

```python
# better dataset
fig, ax = plt.subplots(nrows=2,ncols=2,figsize=(10,10))

ax[0,0].scatter(batters['avg'],batters['strike_rate'],color='red')
ax[0,1].scatter(batters['avg'],batters['runs'])
ax[1,0].hist(batters['avg'])
ax[1,1].hist(batters['runs'])

ax[0,0].set_title('Avarage Vs Strike rate')
ax[0,0].set_ylabel('Strike rate')
ax[0,0].set_xlabel('Avrage')
ax[0,1].set_title('Avarage Vs Runs')
ax[0,1].set_ylabel('Runs')
ax[0,1].set_xlabel('Average')

ax[1,0].set_title('Avarage Histro')
ax[1,1].set_title('Runs Histro')
```

```python
fig = plt.figure()
ax1 = fig.add_subplot(2,2,1)
ax1.scatter(batters['avg'],batters['strike_rate'],color='red')
ax2 = fig.add_subplot(2,2,2)
ax2.hist(batters['avg'])
ax3 = fig.add_subplot(2,2,3)
ax3.hist(batters['runs'])
ax4 = fig.add_subplot(2,2,4)
ax4.scatter(batters['avg'],batters['runs'])
```

# Serface Plot

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
```

```python
xx,yy =np.meshgrid(x,y)
```

```python
z = xx**2+ yy**2
z.shape
```

```python
fig = plt.figure(figsize=(15,7))
ax = plt.subplot(projection='3d')
p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
zo = np.sin(xx) + np.cos(yy)
fig = plt.figure(figsize=(15,7))
ax = plt.subplot(projection='3d')
p = ax.plot_surface(xx,yy,zo,cmap='jet')
fig.colorbar(p)
```

```python
fig = plt.figure(figsize=(15,7))
ax = plt.subplot(projection='3d')
p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

# countuor plot
## 3D to 2D


```python
fig = plt.figure(figsize=(15,7))
ax = plt.subplot()
p = ax.contour(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
# contourf varianat of contour
fig = plt.figure(figsize=(15,7))
ax = plt.subplot()
p = ax.contourf(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
zo = np.sin(xx) + np.cos(yy)
fig = plt.figure(figsize=(15,7))
ax = plt.subplot()
p = ax.contourf(xx,yy,zo,cmap='viridis')
fig.colorbar(p)
```

# Heatmaps

```python
delivery = pd.read_csv('https://drive.google.com/uc?export=download&id=1Lrwi1_9_tzUrULo16PCqE6fAIJNaG2eS')
delivery.head(5)
```

```python
temp_df = delivery[(delivery.ballnumber.isin([1,2,3,4,5,6])) & (delivery.batsman_run == 6)]
temp_df.head(5)
```

```python
grid = temp_df.pivot_table(index = 'overs',columns='ballnumber',values='batsman_run',aggfunc='count')
grid
```

```python
plt.figure(figsize=(15,7))
plt.imshow(grid)
plt.yticks(delivery.overs.unique(),list(range(1,21)))
plt.xticks(np.arange(0,6),list(range(1,7)))
plt.colorbar()
```
