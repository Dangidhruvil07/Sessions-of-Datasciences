### `Q-1:` Using `groupby` make groups using the `"Pclass"` column and find out the average age and total number of missing values in the `"Age"` column for every group.

````python
# code he
pclass_group = titanic_df.groupby("Pclass")
for group in list(pclass_group.groups.keys()):
    print(f"pclass {group}:AVG Age {pclass_group.get_group(group)['Age'].mean()}and total missing values in Age column {pclass_group.get_group(group)['Age'].isna().sum()}")
````

### `Q-2:` Using `groupby` make groups using the `"Pclass"` column and fill every group's `"Embarked"` column's missing values with the mode value of that group. After that, print every group's `"Embarked"` column's value counts in ascending order.

````python
for group in list(pclass_group.groups.keys()):
    mode = pclass_group.get_group(group)['Embarked'].mode()[0]
    pclass_group.get_group(group)['Embarked'].fillna(mode)
    print(pclass_group.get_group(group)['Embarked'].value_counts(ascending  = True))
````

### `Q-3:` Make groups based on `"Embarked"` column. And for each of this embarked group, make another group based on `"Pclass"` and find out the average fare (round off up to 2 decimal places) for each "Pclass" for each group of "Embarked".

**Sample Output:**

```bash
{'C': {1: 105, 2: 25, 3: 11},
 'Q': {1: 90, 2: 12, 3: 11},
 'S': {1: 70, 2: 20, 3: 15}}
```

````python
my_dict = {}
embark_groups = titanic_df.groupby('Embarked')
for embark_group in list(embark_groups.groups.keys()):
    pclass_groups = embark_groups.get_group(embark_group).groupby("Pclass")
    my_dict[embark_group] = {}
    for pclass_group in list(pclass_groups.groups.keys()):
        my_dict[embark_group][pclass_group] = round(pclass_groups.get_group(pclass_group)['Fare'].mean(),2)
my_dict
````

````python
# Columns to normalize
cols_to_perform = [
    "Passes",
    "Passes Completed",
    "Attempted Line Breaks",
    "Completed Line Breaks"
]

# Create a copy of the DataFrame to perform normalization, ensuring all original columns are retained
normalized_df = fifa_df.copy()


normalized_df[cols_to_perform] = (
    normalized_df.groupby("Team")[cols_to_perform]
    .transform(lambda x: (x - x.mean()) / x.std())
)

# Required statistics
result = normalized_df.groupby("Team").agg({
    "Passes": ["min", "max"],
    "Yellow Cards": ["min", "max", "mean"],
    "Attempted Line Breaks": ["min", "max", "std"],
    "Possession (%)": "mean"
})

print(result)
````

###`Q-5:` Find batsman in below category-
* Highest score while chasing
* Best Strike rate while chasing and have faced 100+ balls


> Chasing mean team batting in second inning

````python
df = matches[matches.innings == 2]
df.groupby(['ID',"batter"]).sum().sort_values(by = 'batsman_run',ascending = False).head(1)
````

````python
temp_df = df[~(df.extra_type == 'wides')]
temp_df = temp_df.groupby("batter").agg(
    {
        "batsman_run":"sum",
        "ballnumber":"count",

    }
)
temp_df["strike_rate"] = temp_df["batsman_run"]/temp_df["ballnumber"]*100
temp_df[temp_df['ballnumber']>=1000].sort_values(by = 'strike_rate',ascending = False).reset_index()
````

###`Q-6` Most Successful bowler against any batsman. Find that pair of bowler and batsman.
> Most Successful in terms of dissmissal. A bowler who have dissmissed any batsman most no of times. If any two pairs have same no of dissmisal, consider runs conceded by bowler to that batsman. Those who have concede lesser runs is more successful.

````python
# code here
matches['IsbatterOut'] = matches.batter == matches.player_out
matches.groupby(['bowler','batter']).agg({
    "IsbatterOut":"sum",
    "batsman_run":"sum"
}).sort_values(by = ['IsbatterOut','batsman_run'],ascending = [False,True]).head()
````

###`Q-7`: Most successful batting pair in IPL. Batting pair who have scored most runs playing together.


````python
# code here
# matches.groupby(['batter','non-striker']).agg({
#     "batsman_run":"sum"
# }).sort_values(by = 'batsman_run',ascending = False).head()
def func(x):
  return '-'.join(list(np.sort(x.values)))
matches['batter-pair'] = matches[["batter","non-striker"]].apply(func,axis = 1)
matches.groupby("batter-pair")['total_run'].sum().sort_values(ascending = False).head()
````

###`Q-8:` Make a dataframe for all batting pairs played together.
```
Batsman1 Batsman2 Runs Avg StrikeRate
```

> Just to ease this question you can count wide-balls for strike rate.

````python
# code here
temp_df = matches.groupby(['batter-pair']).agg({
    "total_run":"sum",
    "ballnumber":"count",
    "isWicketDelivery":"sum"
}).reset_index()
temp_df["Batsman 1"] = temp_df["batter-pair"].apply(lambda x:x.split('-')[0])
temp_df["Batsman 2"] = temp_df["batter-pair"].apply(lambda x:x.split('-')[1])
temp_df["Avg"] = temp_df["total_run"]/temp_df["isWicketDelivery"]
temp_df["Strike Rate"] = temp_df["total_run"]/temp_df["ballnumber"]*100
temp_df.sort_values("Strike Rate",ascending = False)
temp_df[["Batsman 1","Batsman 2","total_run","Avg","Strike Rate"]]
````
