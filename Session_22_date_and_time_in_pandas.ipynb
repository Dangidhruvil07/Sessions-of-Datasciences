## why separate objects to handle data and time when python already has datetime functionality



*   syntax wise datetime is very convenient.
*   But the perfoemance takes a hit whilw working with huge data.list Vs numpy array.
*   The weekness of python's datetime format inspired the numpy team to add a set of native time series data type to numpy.
*   The datetime64 dtype encodes dates as 64-bit intengers,add thus allows array of dates to be represented by very compactly.




```python
date = np.array('2015-07-04',dtype = np.datetime64)
date
```

```python
date  + np.arange(12)
```

*  Beacuse of the uniform type of numpy datetime64 arrays,this type of operation can be accomplished much more quickly then if we were working directly with Python's datetime objects , especially as arrays get large
*  Pandas Timestamp object combines the ease-of-use of python datetime with the efficient storage and vectorized interface of numpy.datetime64
*  From a group of these Timestamp objects, pandas can construct of DatetimeIndex that can be used to index data in series or DataFrame

```python
# using python datetime object
pd.DatetimeIndex([dt.datetime(2023,1,1),dt.datetime(2022,1,1),dt.datetime(2021,1,1)])
```

```python
# using datetimeindex as series index
pd.Series([1,2,3],index = dt_index)
```

#  Date_Range Function
 here D is denoted for giving date  i think and also using 2D or like you can do like this 3 date after things as slicing

```python
 # genrate daily dates in a given range
pd.date_range(start='2026/08/30',end = '2026/09/10',freq='D')
```

```python
# alternet days in a given range it gives string date then after 3rd date like 30 , 31 X, 1 X , 2like this it gives
pd.date_range(start='2026/08/30',end = '2026/09/10',freq='3D')
```

```python
 # B -> Business days
pd.date_range(start='2026/08/30',end = '2026/09/10',freq='B')
```

```python
#w -> one week per day
pd.date_range(start='2026/08/30',end = '2026/09/10',freq='W-THU')
```

```python
#H -> Hourly data(factor)
pd.date_range(start='2026/08/30',end = '2026/09/10',freq='6h')
```

```python
#ME -> Month end
pd.date_range(start='2026/08/30',end = '2026/09/10',freq='ME')
```

```python
#MS -> Month Starting
pd.date_range(start='2026/08/30',end = '2026/09/10',freq='MS')
```

```python
# YE -> Year end
pd.date_range(start='2026/08/30',end = '2026/09/10',freq='YE')
```

```python
# using periodes(number of results)
pd.date_range(start='2026/08/30',freq='h',periods=24)
```

to_datetime function

*  converts an exitsing objects to pandas timestamp/datetimeindex object

```python
# simple series example
s = pd.Series(['2023/01/01','2022/01/01','2021/01/01'])
pd.to_datetime(s).dt.day_name()
```

```python
# with errors
s = pd.Series(['2023/01/01','2022/01/01','2021/0100/01'])
pd.to_datetime(s,errors='coerce').dt.day_name()
```
